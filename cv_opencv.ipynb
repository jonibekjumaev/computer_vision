```python
import cv2
```

```python
from google.colab.patches import cv2_imshow
image = cv2.imread('/content/nadezhda-nikolaenko-UONij8U_oVc-unsplash.jpg')
cv2_imshow(image)
```

```python
print('shape:', image.shape)
# 1200 => height
# 1200 => width
# 3 => color channels (RGB) Red-Green-Blue  opencv=(BGR) Blue-Green-Red
print('HEIGHT:', image.shape[0])
print('WIDTH:', image.shape[1])
print('CHANNELS:', image.shape[2])
```

```python
(b, g, r) = image[600,60]
print('Blue:', b, 'Green:', g, 'Red:', r)
```

```python
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
cv2_imshow(gray)
```

```python
resized = cv2.resize(gray, (600, 600))
cv2_imshow(resized)
```

```python
cropped = gray[1500:2500, 500:1500]
cv2_imshow(cropped)
```

```python
rotated = cv2.rotate(gray, cv2.ROTATE_90_CLOCKWISE)
cv2_imshow(rotated)
```

```python
flipped = cv2.flip(gray, 1)
cv2_imshow(flipped)
```

```python
small = cv2.resize(flipped, None, fx=0.2, fy=0.2)  # 5 marta kichraytirish
cv2_imshow(small)
```

```python
cv2.imwrite('small_cat.jpg', small)
```
